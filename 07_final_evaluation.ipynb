# 07 — Final Evaluation

## Objective

This notebook defines the final evaluation protocol for the brain MRI
classification project.

Previous experiments showed that strict group-held-out evaluation causes
a substantial performance drop and that augmentation, group weighting,
and group-aware sampling do not fully resolve this issue.

The objective here is not to further optimize the group-held-out setting,
but to define a defensible primary evaluation protocol that:

1. prevents data leakage;
2. preserves class stratification;
3. evaluates performance on unseen images;
4. keeps strict group-held-out performance as a separate generalization
   stress test.

This notebook represents the final model evaluation stage of the project.
